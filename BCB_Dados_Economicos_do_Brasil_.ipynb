# VOLUME SERVIÇOS

### Volume Serviços

```python
import requests
import pandas as pd
from io import StringIO

# URL da API para o código 23982
url = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.23982/dados?formato=json"

# Fazer a requisição
response = requests.get(url)

# Verificar o status da resposta
if response.status_code == 200:
    # Converter os dados para um DataFrame usando StringIO
    VolumeServices = pd.read_json(StringIO(response.text))
    VolumeServices['data'] = pd.to_datetime(VolumeServices['data'], format='%d/%m/%Y')
    VolumeServices.set_index('data', inplace=True)
    
    # Exibir as datas mais recentes
    print("Dados mais recentes:")
    print(VolumeServices.tail())  # Exibe as últimas 5 linhas do DataFrame
else:
    print(f"Erro ao acessar a API: {response.status_code}")
```

```python
import plotly.express as px


fig = px.line(VolumeServices, x=VolumeServices.index, y=VolumeServices.iloc[:, 0], title='Volume Services Monthly % var.')

fig.update_xaxes(range=['2010-01-01', VolumeServices.index.max()])
fig.update_traces(line_color='#00008B')  # '#00008B' is the hex code for dark blue

fig.show()
```

### Volume Serviços MoM

```python
import plotly.express as px
import pandas as pd

# Calculate MoM percentage change
VolumeServices_MoM = VolumeServices.iloc[:, 0].pct_change() * 100

# Create a DataFrame for plotting
mom_data = pd.DataFrame({'MoM Change (%)': VolumeServices_MoM}, index=VolumeServices.index)

fig = px.bar(mom_data, x=mom_data.index, y='MoM Change (%)', title='Volume Services MoM (%)')

# Add percentage values as text annotations
fig.update_traces(texttemplate='%{y:.2f}%', textposition='outside')

# Set start date (replace '2020-01-01' with your desired start date)
fig.update_xaxes(range=['2022-01-01', mom_data.index.max()])

fig.show()
```

# IPCA BENS YoY E IPCA SERVIÇOS YoY

```python
import requests
import pandas as pd
from io import StringIO
import plotly.express as px

# URLs da API para os códigos 10844 (IPCA Serviços) e 10843 (IPCA Bens)
url_servicos = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.10844/dados?formato=json"
url_bens = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.10843/dados?formato=json"

# Fazer as requisições
response_servicos = requests.get(url_servicos)
response_bens = requests.get(url_bens)

# Verificar o status das respostas
if response_servicos.status_code == 200 and response_bens.status_code == 200:
    # Converter os dados para DataFrames usando StringIO
    IPCASERVICOS = pd.read_json(StringIO(response_servicos.text))
    IPCABENS = pd.read_json(StringIO(response_bens.text))
    
    # Converter as datas para o formato datetime e definir como índice
    IPCASERVICOS['data'] = pd.to_datetime(IPCASERVICOS['data'], format='%d/%m/%Y')
    IPCASERVICOS.set_index('data', inplace=True)
    IPCABENS['data'] = pd.to_datetime(IPCABENS['data'], format='%d/%m/%Y')
    IPCABENS.set_index('data', inplace=True)
    
    # Calcular a média móvel de 12 meses para IPCASERVICOS
    IPCASERVICOS_MA12 = IPCASERVICOS['valor'].rolling(window=12).mean()
    
    # Calcular a média móvel de 12 meses para IPCABENS
    IPCABENS_MA12 = IPCABENS['valor'].rolling(window=12).mean()
    
    # Criar um DataFrame combinado para plotagem
    combined_data = pd.DataFrame({
        'IPCASERVICOS MA12': IPCASERVICOS_MA12,
        'IPCABENS MA12': IPCABENS_MA12
    }, index=IPCASERVICOS.index)  # Usar o índice de IPCASERVICOS
    
    # Criar o gráfico de linhas
    fig = px.line(combined_data, x=combined_data.index,
                  y=['IPCASERVICOS MA12', 'IPCABENS MA12'],
                  title='IPCA Serviços e IPCA Bens (Média Móvel de 12 Meses)')
    
    # Atualizar as cores das linhas
    fig.update_traces(selector=dict(name='IPCASERVICOS MA12'), line_color='#FF8C00')  # Laranja escuro para IPCASERVICOS
    fig.update_traces(selector=dict(name='IPCABENS MA12'), line_color='#00008B')  # Azul para IPCABENS
    
    # Adicionar uma linha horizontal em y=0
    fig.add_hline(y=0, line_width=1, line_dash="solid", line_color="black")
    
    # Mostrar o gráfico
    fig.show()
else:
    print(f"Erro ao acessar a API: Serviços ({response_servicos.status_code}), Bens ({response_bens.status_code})")
```

### IGP-M (MoM) e IPCA (YoY)

```python
import requests
import pandas as pd
from io import StringIO
import plotly.express as px

# URLs da API para os códigos 189 (IGP-M) e 13522 (IPCA YoY)
url_igpm = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.189/dados?formato=json"
url_ipcayoy = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.13522/dados?formato=json"

# Fazer as requisições
response_igpm = requests.get(url_igpm)
response_ipcayoy = requests.get(url_ipcayoy)

# Verificar o status das respostas
if response_igpm.status_code == 200 and response_ipcayoy.status_code == 200:
    # Converter os dados para DataFrames usando StringIO
    IGPM = pd.read_json(StringIO(response_igpm.text))
    IPCAYOY = pd.read_json(StringIO(response_ipcayoy.text))
    
    # Converter as datas para o formato datetime e definir como índice
    IGPM['data'] = pd.to_datetime(IGPM['data'], format='%d/%m/%Y')
    IGPM.set_index('data', inplace=True)
    IPCAYOY['data'] = pd.to_datetime(IPCAYOY['data'], format='%d/%m/%Y')
    IPCAYOY.set_index('data', inplace=True)
    
    # Calcular a média móvel de 12 períodos para o IGP-M
    IGPM_MA12 = IGPM['valor'].rolling(window=12).mean()
    
    # Criar um DataFrame combinado para plotagem
    combined_data = pd.DataFrame({
        'IGP-M (MA12)': IGPM_MA12,  # IGP-M com média móvel de 12 períodos
        'IPCA (YoY)': IPCAYOY['valor']  # IPCA YoY
    }, index=IGPM.index)  # Usar o índice do IGP-M
    
    # Criar o gráfico de linhas
    fig = px.line(combined_data,
                  x=combined_data.index,
                  y=['IGP-M (MA12)', 'IPCA (YoY)'],  # Plota ambas as séries
                  title='IGP-M (MA12) e IPCA (YoY)')
    
    # Personalizar as cores das linhas
    fig.update_traces(selector=dict(name='IGP-M (MA12)'), line_color='#00008B')  # Azul para IGP-M
    fig.update_traces(selector=dict(name='IPCA (YoY)'), line_color='#FF8C00')    # Laranja para IPCA YoY
    
    # Adicionar uma linha horizontal preta em y=0
    fig.add_hline(y=0, line_width=1, line_color="black")
    
    # Mostrar o gráfico
    fig.show()
else:
    print(f"Erro ao acessar a API: IGP-M ({response_igpm.status_code}), IPCA YoY ({response_ipcayoy.status_code})")
```

```python
import plotly.express as px
import pandas as pd
import numpy as np

# Assuming RESULTADOPRIMARIOPIB and TAXADESEMPREGO are pandas DataFrames

# Create a combined DataFrame for plotting
combined_data = pd.DataFrame({
    'RESULTADOPRIMARIOPIB': RESULTADOPRIMARIOPIB.iloc[:, 0],
    'TAXADESEMPREGO': TAXADESEMPREGO.iloc[:, 0]
}, index=RESULTADOPRIMARIOPIB.index)

# Calculate z-scores for each series
for column in combined_data.columns:
    combined_data[column] = (combined_data[column] - combined_data[column].mean()) / combined_data[column].std()

# Create the line plot
fig = px.line(combined_data, x=combined_data.index,
              y=['RESULTADOPRIMARIOPIB', 'TAXADESEMPREGO'],
              title='Resultado Primário/PIB e Taxa de Desemprego (Z-scores)')

# Update line colors
fig.update_traces(selector=dict(name='RESULTADOPRIMARIOPIB'), line_color='#00008B')  # Blue for RESULTADOPRIMARIOPIB
fig.update_traces(selector=dict(name='TAXADESEMPREGO'), line_color='#FF8C00')  # Dark orange for TAXADESEMPREGO

# Set start date (replace '2007-01-01' with your desired start date)
fig.update_xaxes(range=['2007-01-01', combined_data.index.max()])

# Add a horizontal line at y = 0
fig.add_hline(y=0, line_width=1, line_dash="solid", line_color="black")

fig.show()
```

### Saldo PJ YoY, Saldo PF YoY, Saldo Crédito YoY

```python
import requests
import pandas as pd
from io import StringIO
import plotly.express as px

# URLs da API para os códigos 20540 (SALDO CRÉDITO PJ), 20541 (SALDO CRÉDITO PF) e 20539 (SALDO CRÉDITO TOTAL)
url_pj = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.20540/dados?formato=json"
url_pf = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.20541/dados?formato=json"
url_total = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.20539/dados?formato=json"

# Fazer as requisições
response_pj = requests.get(url_pj)
response_pf = requests.get(url_pf)
response_total = requests.get(url_total)

# Verificar o status das respostas
if response_pj.status_code == 200 and response_pf.status_code == 200 and response_total.status_code == 200:
    # Converter os dados para DataFrames usando StringIO
    SALDOCREDITOPJ = pd.read_json(StringIO(response_pj.text))
    SALDOCREDITOPF = pd.read_json(StringIO(response_pf.text))
    SALDOCREDITOTOTAL = pd.read_json(StringIO(response_total.text))
    
    # Converter as datas para o formato datetime e definir como índice
    SALDOCREDITOPJ['data'] = pd.to_datetime(SALDOCREDITOPJ['data'], format='%d/%m/%Y')
    SALDOCREDITOPJ.set_index('data', inplace=True)
    SALDOCREDITOPF['data'] = pd.to_datetime(SALDOCREDITOPF['data'], format='%d/%m/%Y')
    SALDOCREDITOPF.set_index('data', inplace=True)
    SALDOCREDITOTOTAL['data'] = pd.to_datetime(SALDOCREDITOTOTAL['data'], format='%d/%m/%Y')
    SALDOCREDITOTOTAL.set_index('data', inplace=True)
    
    # Calcular a variação percentual YoY para cada série
    SALDOCREDITOPJ_YoY = SALDOCREDITOPJ['valor'].pct_change(periods=12) * 100
    SALDOCREDITOPF_YoY = SALDOCREDITOPF['valor'].pct_change(periods=12) * 100
    SALDOCREDITOTOTAL_YoY = SALDOCREDITOTOTAL['valor'].pct_change(periods=12) * 100
    
    # Criar um DataFrame combinado para plotagem
    combined_yoy_data = pd.DataFrame({
        'SALDOCREDITOPJ_YoY': SALDOCREDITOPJ_YoY,
        'SALDOCREDITOPF_YoY': SALDOCREDITOPF_YoY,
        'SALDOCREDITOTOTAL_YoY': SALDOCREDITOTOTAL_YoY
    }, index=SALDOCREDITOPJ.index)  # Usar o índice de SALDOCREDITOPJ
    
    # Criar o gráfico de linha
    fig = px.line(combined_yoy_data, x=combined_yoy_data.index,
                  y=['SALDOCREDITOPJ_YoY', 'SALDOCREDITOPF_YoY', 'SALDOCREDITOTOTAL_YoY'],
                  title='Saldo de Crédito PJ, PF e Total YoY')
    
    # Atualizar as cores das linhas
    fig.update_traces(selector=dict(name='SALDOCREDITOPJ_YoY'), line_color='blue')       # Azul para SALDOCREDITOPJ
    fig.update_traces(selector=dict(name='SALDOCREDITOPF_YoY'), line_color='darkorange') # Laranja escuro para SALDOCREDITOPF
    fig.update_traces(selector=dict(name='SALDOCREDITOTOTAL_YoY'), line_color='gray')    # Cinza para SALDOCREDITOTOTAL
    
    # Configurar a data inicial (um ano após a data inicial para evitar valores NaN)
    fig.update_xaxes(range=['2008-01-01', combined_yoy_data.index.max()])
    
    # Mostrar o gráfico
    fig.show()
else:
    print(f"Erro ao acessar a API: PJ ({response_pj.status_code}), PF ({response_pf.status_code}), Total ({response_total.status_code})")
```

### Índice de Custo de Crédito (ICC) e Selic Mensal Atualizada

```python
import requests
import pandas as pd
from io import StringIO
import plotly.express as px

# URLs da API para os códigos 25351 (Índice de Custo de Crédito - ICC) e 4189 (SELIC Mensal)
url_icc = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.25351/dados?formato=json"
url_selic = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.4189/dados?formato=json"

# Fazer as requisições
response_icc = requests.get(url_icc)
response_selic = requests.get(url_selic)

# Verificar o status das respostas
if response_icc.status_code == 200 and response_selic.status_code == 200:
    # Converter os dados para DataFrames usando StringIO
    IndiceDeCustoDeCredito = pd.read_json(StringIO(response_icc.text))
    SELICMENSAL = pd.read_json(StringIO(response_selic.text))
    
    # Converter as datas para o formato datetime e definir como índice
    IndiceDeCustoDeCredito['data'] = pd.to_datetime(IndiceDeCustoDeCredito['data'], format='%d/%m/%Y')
    IndiceDeCustoDeCredito.set_index('data', inplace=True)
    SELICMENSAL['data'] = pd.to_datetime(SELICMENSAL['data'], format='%d/%m/%Y')
    SELICMENSAL.set_index('data', inplace=True)
    
    # Criar um DataFrame combinado para plotagem
    combined_data = pd.DataFrame({
        'IndiceDeCustoDeCredito': IndiceDeCustoDeCredito['valor'],
        'SELICMENSAL': SELICMENSAL['valor']
    }, index=IndiceDeCustoDeCredito.index)
    
    # Normalizar os dados usando z-scores
    for column in combined_data.columns:
        combined_data[column] = (combined_data[column] - combined_data[column].mean()) / combined_data[column].std()
    
    # Criar o gráfico de linha
    fig = px.line(combined_data, x=combined_data.index,
                  y=['IndiceDeCustoDeCredito', 'SELICMENSAL'],
                  title='Índice de Custo de Crédito (ICC) e Selic Mensal Atualizada')
    
    # Atualizar as cores das linhas
    fig.update_traces(selector=dict(name='IndiceDeCustoDeCredito'), line_color='blue')       # Azul para ICC
    fig.update_traces(selector=dict(name='SELICMENSAL'), line_color='darkorange')           # Laranja escuro para SELIC
    
    # Configurar a data inicial
    fig.update_xaxes(range=['2007-01-01', combined_data.index.max()])
    
    # Adicionar uma linha horizontal em y = 0
    fig.add_hline(y=0, line_width=1, line_dash="solid", line_color="black")
    
    # Mostrar o gráfico
    fig.show()
else:
    print(f"Erro ao acessar a API: ICC ({response_icc.status_code}), SELIC ({response_selic.status_code})")
```
