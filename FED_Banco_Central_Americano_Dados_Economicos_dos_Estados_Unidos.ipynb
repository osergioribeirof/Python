### New House Sales e 30Y Mortgage Rate (Average)

```python
import pandas_datareader.data as web
import datetime
import plotly.graph_objects as go
from scipy.stats import zscore

start = datetime.datetime(2010, 1, 1)

# Fetch the data
NEWHOUSESALES = web.DataReader('HSN1F', 'fred', start)
MortgageRate = web.DataReader('MORTGAGE30US', 'fred', start)

# Normalize the data using z-scores
NEWHOUSESALES['HSN1F_zscore'] = zscore(NEWHOUSESALES['HSN1F'])
MortgageRate['MORTGAGE30US_zscore'] = zscore(MortgageRate['MORTGAGE30US'])

# Create the figure
fig = go.Figure()

# Add New House Sales trace (blue)
fig.add_trace(go.Scatter(
    x=NEWHOUSESALES.index,
    y=NEWHOUSESALES['HSN1F_zscore'],  # Using normalized data
    mode='lines',
    name='New House Sales',
    line=dict(color='blue')
))

# Add Mortgage Rate trace (dark orange)
fig.add_trace(go.Scatter(
    x=MortgageRate.index,
    y=MortgageRate['MORTGAGE30US_zscore'],  # Using normalized data
    mode='lines',
    name='Mortgage Rate',
    line=dict(color='darkorange')
))

# Update layout
fig.update_layout(
    title='New House Sales and Mortgage Rate (Normalized)',
    xaxis_title='Data',
    yaxis_title='Z-score'
)

# Show the figure
fig.show()
```

 S&P 500 Case Shiller YoY

```python
import pandas_datareader.data as web
import datetime
import plotly.graph_objects as go
import numpy as np
from scipy.stats import zscore

start = datetime.datetime(2000, 1, 1)
CaseShiller = web.DataReader('CSUSHPINSA', 'fred', start)

# Calcula a variação percentual anual (YoY)
CaseShiller['YoY_Change'] = CaseShiller['CSUSHPINSA'].pct_change(periods=12) * 100

# Normaliza os dados usando zscore
CaseShiller['CSUSHPINSA_zscore'] = zscore(CaseShiller['CSUSHPINSA'])
CaseShiller['YoY_Change_zscore'] = zscore(CaseShiller['YoY_Change'])

# Cria o gráfico
fig = go.Figure()

# Adiciona o traço para o CaseShiller normalizado - Azul
fig.add_trace(go.Scatter(x=CaseShiller.index,
                         y=CaseShiller['CSUSHPINSA_zscore'],
                         mode='lines', name='CaseShiller (Normalizado)',
                         line=dict(color='blue')))

# Adiciona o traço para o CaseShiller em YoY - Laranja
fig.add_trace(go.Scatter(x=CaseShiller.index, y=CaseShiller['YoY_Change'],
                         mode='lines', name='CaseShiller YoY',
                         line=dict(color='darkorange')))

# Adiciona linha preta no 0
fig.add_shape(type='line',
              x0=CaseShiller.index.min(),
              y0=0,
              x1=CaseShiller.index.max(),
              y1=0,
              line=dict(color='black'))

# Configura o layout do gráfico
fig.update_layout(title='CaseShiller e CaseShiller YoY (Normalizados)',
                  xaxis_title='Data',
                  yaxis_title='Z-score')

# Exibe o gráfico
fig.show()
```

```python
import pandas_datareader.data as web
import datetime
import plotly.express as px

start = datetime.datetime(2018, 1, 1)
REALGDP = web.DataReader('GDPC1', 'fred', start)

# Calcula a variação percentual trimestral anualizada (YoY)
REALGDP['YoY_Change'] = REALGDP['GDPC1'].pct_change(periods=4) * 100
REALGDP['Quarter'] = REALGDP.index.to_period('Q')  # Cria uma coluna com o trimestre

# Agrupa por trimestre e calcula a média do YoY Change
quarterly_gdp = REALGDP.groupby('Quarter')['YoY_Change'].mean().reset_index()

# Converte a coluna 'Quarter' para string
quarterly_gdp['Quarter'] = quarterly_gdp['Quarter'].astype(str)

# Cria o gráfico de barras com Plotly Express
fig_yoy = px.bar(quarterly_gdp, x='Quarter', y='YoY_Change', title='REALGDP YoY Trimestral',
                  color_discrete_sequence=['blue'])
fig_yoy.update_traces(texttemplate='%{y:.2f}%', textposition='outside') # valor acima
fig_yoy.update_layout(xaxis_title='Trimestre', yaxis_title='Variação Percentual Anual Média (%)')
fig_yoy.show()
```

Real Final Sales to Private DPP QoQ Annualized

```python
import pandas as pd
import datetime
import plotly.express as px
from pandas_datareader import data as web

# Coletar dados
start = datetime.datetime(2017, 1, 1)
FinalSalesToPrivateDomestic = web.DataReader('LB0000031Q020SBEA', 'fred', start)

# Calcular a variação percentual trimestral anualizada (QoQ Annualized)
FinalSalesToPrivateDomestic['QoQ_Annualized'] = (
    (FinalSalesToPrivateDomestic['LB0000031Q020SBEA'] / FinalSalesToPrivateDomestic['LB0000031Q020SBEA'].shift(1) - 1) * 4
) * 100  # Multiplicar por 100 para obter a porcentagem

# Converter o índice para data trimestral
FinalSalesToPrivateDomestic.index = FinalSalesToPrivateDomestic.index.to_period('Q')

# Converter o índice para string antes de plotar
FinalSalesToPrivateDomestic.index = FinalSalesToPrivateDomestic.index.astype(str)

# Criar uma nova coluna com os valores formatados como porcentagem
FinalSalesToPrivateDomestic['QoQ_Percentual_Text'] = FinalSalesToPrivateDomestic['QoQ_Annualized'].apply(lambda x: f'{x:.2f}%')

# Calcular a média de 'QoQ_Annualized'
media_qoq = FinalSalesToPrivateDomestic['QoQ_Annualized'].mean()

# Criar o histograma
fig = px.histogram(FinalSalesToPrivateDomestic,
                   x=FinalSalesToPrivateDomestic.index,
                   y='QoQ_Annualized',
                   title='Real Final Sales to Private DPP QoQ Annualized & Average Real Final Sales to DPP')

# Customizar o histograma
fig.update_layout(xaxis_title='Data Trimestral',
                  yaxis_title='Variação Percentual (%)')
fig.update_xaxes(type='category')  # Definir o eixo x como categórico para datas trimestrais

# Adicionar texto com fig.update_traces()
fig.update_traces(text=FinalSalesToPrivateDomestic['QoQ_Percentual_Text'], textposition="outside")

# Adicionar a linha da média em laranja
fig.add_shape(type='line',
              x0=FinalSalesToPrivateDomestic.index[0],
              y0=media_qoq,
              x1=FinalSalesToPrivateDomestic.index[-1],
              y1=media_qoq,
              line=dict(color='orange', width=2))

# Exibir o histograma
fig.show()
```

### ECI QoQ e AverageQuarterly ECI

```python
import pandas_datareader.data as web
import datetime
import plotly.express as px
import plotly.graph_objects as go

start = datetime.datetime(2017, 1, 1)
EMPLOYMENTCOST = web.DataReader('ECIALLCIV', 'fred', start)

# Calculate QoQ percentage change
EMPLOYMENTCOST['QoQ_Change'] = EMPLOYMENTCOST['ECIALLCIV'].pct_change() * 100

# Format the percentage change for display
EMPLOYMENTCOST['QoQ_Change_Text'] = EMPLOYMENTCOST['QoQ_Change'].apply(lambda x: f'{x:.2f}%')

# Calculate average quarterly percentage change
average_quarterly_pct_change = EMPLOYMENTCOST['QoQ_Change'].mean()

# Create the bar plot
fig = px.bar(EMPLOYMENTCOST, x=EMPLOYMENTCOST.index, y='QoQ_Change',
             title='Employment Cost Index (ECI) - QoQ Percentage Change',
             text='QoQ_Change_Text',
             color_discrete_sequence=['blue'])

# Add the average quarterly line
fig.add_shape(
    type="line",
    x0=EMPLOYMENTCOST.index.min(),
    y0=average_quarterly_pct_change,
    x1=EMPLOYMENTCOST.index.max(),
    y1=average_quarterly_pct_change,
    line=dict(color="darkorange", width=2),
)

# Customize the layout and text position
fig.update_layout(xaxis_title='Data', yaxis_title='Percentage Change (%)')
fig.update_traces(textposition='outside', selector=dict(type='bar'))  # Apply only to bar traces

# Format x-axis labels to quarters
fig.update_xaxes(
    tickmode='array',
    tickvals=EMPLOYMENTCOST.index,
    ticktext=EMPLOYMENTCOST.index.to_series().dt.to_period("Q").astype(str)
)

# Display the plot
fig.show()
```

### Retail Money Market Funds (TRI) E Fed Funds Rate (FFR)

```python
import pandas_datareader.data as web
import datetime
import plotly.graph_objects as go
from scipy.stats import zscore

start = datetime.datetime(2000, 1, 1)

# Fetch the data
FFR = web.DataReader('DFF', 'fred', start)
REALMONEYMARKETFUNDS = web.DataReader('WRMFNS', 'fred', start)

# Normalize the data using z-scores
FFR['DFF_zscore'] = zscore(FFR['DFF'])
REALMONEYMARKETFUNDS['WRMFNS_zscore'] = zscore(REALMONEYMARKETFUNDS['WRMFNS'])

# Create the figure
fig = go.Figure()

# Add FFR trace (blue)
fig.add_trace(go.Scatter(
    x=FFR.index,
    y=FFR['DFF_zscore'],  # Using normalized data
    mode='lines',
    name='FFR',
    line=dict(color='blue')
))

# Add REALMONEYMARKETFUNDS trace (orange)
fig.add_trace(go.Scatter(
    x=REALMONEYMARKETFUNDS.index,
    y=REALMONEYMARKETFUNDS['WRMFNS_zscore'],  # Using normalized data
    mode='lines',
    name='REALMONEYMARKETFUNDS',
    line=dict(color='darkorange')
))

# Update layout
fig.update_layout(
    title='FFR vs REALMONEYMARKETFUNDS (Normalized)',
    xaxis_title='Data',
    yaxis_title='Z-score'
)

# Show the figure
fig.show()
```
