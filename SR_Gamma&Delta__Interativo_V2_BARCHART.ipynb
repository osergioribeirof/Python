### GAMMA

```python
# Black-Scholes European-Options Gamma
def calcGammaEx(S, K, vol, T, r, q, optType, OI):
    if T == 0 or vol == 0:
        return 0

    dp = (np.log(S/K) + (r - q + 0.5*vol**2)*T) / (vol*np.sqrt(T))
    dm = dp - vol*np.sqrt(T)

    if optType == 'call':
        gamma = np.exp(-q*T) * norm.pdf(dp) / (S * vol * np.sqrt(T))
        return OI * 100 * S * S * 0.01 * gamma
    else: # Gamma is same for calls and puts. This is just to cross-check
        gamma = K * np.exp(-r*T) * norm.pdf(dm) / (S * S * vol * np.sqrt(T))
        return OI * 100 * S * S * 0.01 * gamma

def isThirdFriday(d):
    return d.weekday() == 4 and 15 <= d.day <= 21
```

```python
# ================================================================================
# ### GAMMA FLIP DINÂMICO E ESTRUTURAL
# ================================================================================

# CÉLULA NOVA - Cálculo dos Gamma Flips

# --------------------------------------------------------------------------------
# 1. GAMMA FLIP DINÂMICO
# --------------------------------------------------------------------------------
# Objetivo: Encontrar o preço onde o GEX total muda de sinal (de negativo para
# positivo ou vice-versa) ao simular diferentes preços do ativo.

def calcular_gamma_flip_dinamico(df, spot, r=0.05, q=0):
    """
    Calcula o Gamma Flip Dinâmico recalculando gamma para preços hipotéticos.

    Args:
        df: DataFrame com dados de opções
        spot: Preço atual do ativo
        r: Taxa livre de risco (default 5%)
        q: Dividend yield (default 0%)

    Returns:
        flip_price: Preço onde ocorre o flip
        gex_curve: Array com valores de GEX para cada preço testado
        price_range: Array com preços testados
    """
    # Define range de preços para testar (0.8 a 1.2 do spot)
    price_range = np.linspace(0.8 * spot, 1.2 * spot, 100)
    gex_total = []

    # Calcula tempo até expiração (em anos)
    expiration = df['ExpirationDate'].iloc[0]
    T = (expiration - datetime.now()).days / 365.0

    if T <= 0:
        print("ATENÇÃO: Opções já expiraram!")
        return None, None, None

    # Para cada preço hipotético, recalcula gamma e soma GEX
    for S_hyp in price_range:
        gex_calls = 0
        gex_puts = 0

        for idx, row in df.iterrows():
            K = row['StrikePrice']

            # Recalcula gamma para calls
            if row['CallIV'] > 0 and row['CallOpenInt'] > 0:
                vol_call = row['CallIV']
                gamma_call_recalc = calcGammaEx(S_hyp, K, vol_call, T, r, q, 'call', row['CallOpenInt'])
                gex_calls += gamma_call_recalc

            # Recalcula gamma para puts
            if row['PutIV'] > 0 and row['PutOpenInt'] > 0:
                vol_put = row['PutIV']
                gamma_put_recalc = calcGammaEx(S_hyp, K, vol_put, T, r, q, 'put', row['PutOpenInt'])
                gex_puts += gamma_put_recalc

        # GEX total = calls positivos - puts (que já são multiplicados por -1)
        gex_total.append((gex_calls - gex_puts) / 10**9)

    gex_total = np.array(gex_total)

    # Encontra onde o sinal muda
    signs = np.sign(gex_total)
    sign_changes = np.where(np.diff(signs))[0]

    if len(sign_changes) > 0:
        flip_idx = sign_changes[0]
        flip_price = price_range[flip_idx]
        return flip_price, gex_total, price_range
    else:
        print("Nenhum Gamma Flip Dinâmico encontrado na faixa de preços testada.")
        return None, gex_total, price_range

# --------------------------------------------------------------------------------
# 2. GAMMA FLIP ESTRUTURAL (Zero Gamma por OI)
# --------------------------------------------------------------------------------
# Objetivo: Encontrar o strike onde o GEX acumulado está mais próximo de zero,
# representando o ponto de equilíbrio estrutural do mercado.

def calcular_gamma_flip_estrutural(df, spot):
    """
    Calcula o Gamma Flip Estrutural usando gamma e delta observados no spot real.

    Args:
        df: DataFrame com dados de opções
        spot: Preço atual do ativo

    Returns:
        equilibrium_strike: Strike de equilíbrio
        gex_accumulated: Array com GEX acumulado por strike
    """
    # Agrupa por strike
    df_grouped = df.groupby('StrikePrice').sum(numeric_only=True).reset_index()

    # Calcula GEX_OI para calls e puts
    # Fórmula: γ × OI × Sign(Δ) × 100 × S² × 0.01

    gex_oi_calls = (df_grouped['CallGamma'] *
                    df_grouped['CallOpenInt'] *
                    np.sign(df_grouped['CallDelta']) *
                    100 * spot**2 * 0.01)

    gex_oi_puts = (df_grouped['PutGamma'] *
                   df_grouped['PutOpenInt'] *
                   np.sign(df_grouped['PutDelta']) *
                   100 * spot**2 * 0.01)

    # GEX total por strike (calls - puts)
    gex_oi_total = (gex_oi_calls - gex_oi_puts) / 10**9

    # Acumula o GEX através dos strikes
    gex_accumulated = np.cumsum(gex_oi_total)

    # Encontra o strike onde o acumulado está mais próximo de zero
    zero_idx = np.argmin(np.abs(gex_accumulated))
    equilibrium_strike = df_grouped['StrikePrice'].iloc[zero_idx]

    return equilibrium_strike, gex_accumulated, df_grouped['StrikePrice'].values

# --------------------------------------------------------------------------------
# EXECUÇÃO DOS CÁLCULOS
# --------------------------------------------------------------------------------

print("="*80)
print("CALCULANDO GAMMA FLIPS")
print("="*80)

# Gamma Flip Dinâmico
flip_dinamico, gex_curve, price_range = calcular_gamma_flip_dinamico(df, spotPrice)

if flip_dinamico:
    print(f"\n🎯 GAMMA FLIP DINÂMICO: ${flip_dinamico:,.2f}")
    print(f"   → Market makers mudam de comportamento neste preço")
    print(f"   → Abaixo: tendem a amortecer movimentos (compram queda, vendem alta)")
    print(f"   → Acima: tendem a amplificar movimentos (compram alta, vendem queda)")

# Gamma Flip Estrutural
flip_estrutural, gex_accum, strikes_array = calcular_gamma_flip_estrutural(df, spotPrice)

print(f"\n🏛️  GAMMA FLIP ESTRUTURAL: ${flip_estrutural:,.2f}")
print(f"   → Centro de equilíbrio da estrutura institucional")
print(f"   → Strike onde o book de opções está balanceado")

print(f"\n📍 SPOT ATUAL: ${spotPrice:,.2f}")
print("="*80)

# --------------------------------------------------------------------------------
# VISUALIZAÇÃO
# --------------------------------------------------------------------------------

# Chart: Gamma Flip Dinâmico - Curva de GEX vs Preço
if flip_dinamico and gex_curve is not None:
    fig_dinamico = go.Figure()

    # Linha de GEX
    fig_dinamico.add_trace(go.Scatter(
        x=price_range,
        y=gex_curve,
        mode='lines',
        name='GEX Total',
        line=dict(color='blue', width=2)
    ))

    # Linha horizontal em zero
    fig_dinamico.add_hline(y=0, line_dash="dash", line_color="gray", opacity=0.5)

    # Linha vertical no Gamma Flip
    fig_dinamico.add_vline(
        x=flip_dinamico,
        line_dash="dash",
        line_color="red",
        annotation_text=f"Gamma Flip: ${flip_dinamico:,.0f}",
        annotation_position="top"
    )

    # Linha vertical no Spot
    fig_dinamico.add_vline(
        x=spotPrice,
        line_dash="dot",
        line_color="green",
        annotation_text=f"Spot: ${spotPrice:,.0f}",
        annotation_position="bottom"
    )

    fig_dinamico.update_layout(
        title={
            'text': "Gamma Flip Dinâmico - GEX vs Preço do Ativo",
            'font': {'size': 20, 'family': 'Arial Black'}
        },
        xaxis_title="Preço do Ativo",
        yaxis_title="GEX Total ($ Bn)",
        plot_bgcolor='white',
        font=dict(family='Arial', size=12, color='black'),
        width=1750,
        height=600,
        hovermode='x unified'
    )

    fig_dinamico.show()

# Chart: Gamma Flip Estrutural - GEX Acumulado por Strike
fig_estrutural = go.Figure()

fig_estrutural.add_trace(go.Scatter(
    x=strikes_array,
    y=gex_accum,
    mode='lines+markers',
    name='GEX Acumulado',
    line=dict(color='purple', width=2),
    marker=dict(size=4)
))

# Linha horizontal em zero
fig_estrutural.add_hline(y=0, line_dash="dash", line_color="gray", opacity=0.5)

# Linha vertical no Gamma Flip Estrutural
fig_estrutural.add_vline(
    x=flip_estrutural,
    line_dash="dash",
    line_color="red",
    annotation_text=f"Flip Estrutural: ${flip_estrutural:,.0f}",
    annotation_position="top right"
)

# Linha vertical no Spot
fig_estrutural.add_vline(
    x=spotPrice,
    line_dash="dot",
    line_color="green",
    annotation_text=f"Spot: ${spotPrice:,.0f}",
    annotation_position="bottom left"
)

fig_estrutural.update_layout(
    title={
        'text': "Gamma Flip Estrutural - GEX Acumulado por Strike",
        'font': {'size': 20, 'family': 'Arial Black'}
    },
    xaxis_title="Strike Price",
    yaxis_title="GEX Acumulado ($ Bn)",
    xaxis=dict(range=[fromStrike, toStrike]),
    plot_bgcolor='white',
    font=dict(family='Arial', size=12, color='black'),
    width=1750,
    height=600,
    hovermode='x unified'
)

fig_estrutural.show()

# --------------------------------------------------------------------------------
# EXPORT DOS RESULTADOS
# --------------------------------------------------------------------------------

# Cria DataFrame com os resultados
resultados_flip = pd.DataFrame({
    'Métrica': ['Spot Atual', 'Gamma Flip Dinâmico', 'Gamma Flip Estrutural'],
    'Valor': [spotPrice, flip_dinamico if flip_dinamico else np.nan, flip_estrutural]
})

print("\n📊 RESUMO DOS FLIPS:")
print(resultados_flip.to_string(index=False))
```

```python
# CÉLULA 19
# ---=== CALCULATE GAMMA PROFILE ===---
levels = np.linspace(fromStrike, toStrike, 60)

# For 0DTE options, I'm setting DTE = 1 day, otherwise they get excluded
df['daysTillExp'] = [1/262 if (np.busday_count(todayDate.date(), x.date())) == 0 \
                           else np.busday_count(todayDate.date(), x.date())/262 for x in df.ExpirationDate]

nextExpiry = df['ExpirationDate'].min()

df['IsThirdFriday'] = [isThirdFriday(x) for x in df.ExpirationDate]
thirdFridays = df.loc[df['IsThirdFriday'] == True]
nextMonthlyExp = thirdFridays['ExpirationDate'].min()

totalGamma = []
totalGammaExNext = []
totalGammaExFri = []
```

```python
# CÉLULA 20
# For each spot level, calc gamma exposure at that point
for level in levels:
    df['callGammaEx'] = df.apply(lambda row : calcGammaEx(level, row['StrikePrice'], row['CallIV'],
                                                          row['daysTillExp'], 0, 0, "call", row['CallOpenInt']), axis = 1)

    df['putGammaEx'] = df.apply(lambda row : calcGammaEx(level, row['StrikePrice'], row['PutIV'],
                                                         row['daysTillExp'], 0, 0, "put", row['PutOpenInt']), axis = 1)

    totalGamma.append(df['callGammaEx'].sum() - df['putGammaEx'].sum())

    exNxt = df.loc[df['ExpirationDate'] != nextExpiry]
    totalGammaExNext.append(exNxt['callGammaEx'].sum() - exNxt['putGammaEx'].sum())

    exFri = df.loc[df['ExpirationDate'] != nextMonthlyExp]
    totalGammaExFri.append(exFri['callGammaEx'].sum() - exFri['putGammaEx'].sum())

totalGamma = np.array(totalGamma) / 10**9
totalGammaExNext = np.array(totalGammaExNext) / 10**9
totalGammaExFri = np.array(totalGammaExFri) / 10**9
```

```python
# CÉLULA 21
# Find Gamma Flip Point
zeroCrossIdx = np.where(np.diff(np.sign(totalGamma)))[0]

negGamma = totalGamma[zeroCrossIdx]
posGamma = totalGamma[zeroCrossIdx+1]
negStrike = levels[zeroCrossIdx]
posStrike = levels[zeroCrossIdx+1]

zeroGamma = posStrike - ((posStrike - negStrike) * posGamma/(posGamma-negGamma))
zeroGamma = zeroGamma[0]
```

```python
# ================================================================================
# ### DELTA (CHART 4,5 E 6)
# ================================================================================


# CÉLULA 25
# ---=== CALCULATE SPOT DELTA ===---
# Delta Exposure = Unit Delta * Open Interest * Contract Size * Spot Price
df['CallDEX'] = df['CallDelta'] * df['CallOpenInt'] * 100 * spotPrice
df['PutDEX'] = df['PutDelta'] * df['PutOpenInt'] * 100 * spotPrice

# Total Delta considers the sign of delta for calls and puts
df['TotalDelta'] = (df.CallDEX + df.PutDEX) / 10**6 # Converting to millions for better scaling

dfAgg_delta = df.groupby(['StrikePrice']).sum(numeric_only=True)
strikes_delta = dfAgg_delta.index.values
```

```python
# ---=== CALCULATE DELTA PROFILE ===---
levels_delta = np.linspace(fromStrike, toStrike, 60)

totalDelta = []
totalDeltaExNext = []
totalDeltaExFri = []

# For each spot level, calc delta exposure at that point
for level in levels_delta:
    df['callDeltaEx'] = df.apply(lambda row : row['CallDelta'] * row['CallOpenInt'] * 100 * level, axis = 1)

    df['putDeltaEx'] = df.apply(lambda row : row['PutDelta'] * row['PutOpenInt'] * 100 * level, axis = 1)

    totalDelta.append(df['callDeltaEx'].sum() + df['putDeltaEx'].sum())

    exNxt_delta = df.loc[df['ExpirationDate'] != nextExpiry]
    totalDeltaExNext.append(exNxt_delta['callDeltaEx'].sum() + exNxt_delta['putDeltaEx'].sum())

    exFri_delta = df.loc[df['ExpirationDate'] != nextMonthlyExp]
    totalDeltaExFri.append(exFri_delta['callDeltaEx'].sum() + exFri_delta['putDeltaEx'].sum())

totalDelta = np.array(totalDelta) / 10**6 # Converting to millions
totalDeltaExNext = np.array(totalDeltaExNext) / 10**6 # Converting to millions
totalDeltaExFri = np.array(totalDeltaExFri) / 10**6 # Converting to millions

# Find Delta Flip Point
zeroCrossIdx_delta = np.where(np.diff(np.sign(totalDelta)))[0]

# Handle the case where there is no zero cross
if zeroCrossIdx_delta.size > 0:
    negDelta = totalDelta[zeroCrossIdx_delta]
    posDelta = totalDelta[zeroCrossIdx_delta+1]
    negStrike_delta = levels_delta[zeroCrossIdx_delta]
    posStrike_delta = levels_delta[zeroCrossIdx_delta+1]

    zeroDelta = posStrike_delta - ((posStrike_delta - negStrike_delta) * posDelta/(posDelta-negDelta))
    # Keep zeroDelta as a single value if there's a cross, otherwise set to None or a default
    zeroDelta = zeroDelta[0] if zeroDelta.size > 0 else None
else:
    zeroDelta = None # Set zeroDelta to None if no zero cross is found
```
