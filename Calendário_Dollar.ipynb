### Payroll (Investing.com)

```python
import pandas_datareader.data as web
import datetime
import plotly.express as px  # Import plotly.express for line chart

# Definir a data de início
start = datetime.datetime(2022, 1, 1)

# Coletar dados do Payroll (PAYEMS)
Payroll = web.DataReader('PAYEMS', 'fred', start)

# Calcular a diferença entre o valor atual e o valor do mês passado
Payroll['Diff'] = Payroll['PAYEMS'].diff()

# Multiplicar a diferença por 1000 para obter o novo valor do Payroll
Payroll['PAYEMS'] = Payroll['Diff'] * 1000

# Remover a primeira linha que contém um valor NaN na coluna 'Diff'
Payroll = Payroll.dropna()

# Criar o gráfico de linha com Plotly Express
fig = px.bar(Payroll, x=Payroll.index, y='PAYEMS', title='Payroll', text_auto='.2s')


# Formatar eixo x para mostrar ano e mês
fig.update_xaxes(
    tickformat="%Y-%m",  # Formato para ano e mês (YYYY-MM)
    dtick="M1",          # Intervalo de ticks mensal
    title_text="Data"     # Título do eixo x (opcional)
)

# Configurar layout do gráfico
fig.update_layout(yaxis_title='Valor')
# Change 'clipon' to 'cliponaxis'
fig.update_traces(textfont_size=12, textangle=0, textposition="outside", cliponaxis=False)

# Exibir o gráfico
fig.show()
```

### IPP E Core IPP (Variação Mensal)

```python
#PPI and Core PPI (Monthly Percentage Change) Bar Chart
import pandas_datareader.data as web
import datetime
import plotly.express as px
import pandas as pd

# Definir a data de início e fim
start = datetime.datetime(2023, 1, 1)
end = datetime.datetime(2025, 12, 31)

# Coletar dados do PPI e Core PPI dentro do período especificado
ppi = web.DataReader('PPIFIS', 'fred', start, end)
coreppi = web.DataReader('WPSFD49116', 'fred', start, end)

# Calcular a variação percentual
ppi_percentual = ppi.pct_change()*100
coreppi_percentual = coreppi.pct_change()*100

# Juntar os dados em um único DataFrame para o gráfico
dados_grafico = pd.concat([ppi_percentual.rename(columns={'PPIFIS': 'PPI'}),
                           coreppi_percentual.rename(columns={'WPSFD49116': 'CorePPI'})], axis=1)

# Arredondar os valores para duas casas decimais
dados_grafico['PPI'] = dados_grafico['PPI'].round(2)
dados_grafico['CorePPI'] = dados_grafico['CorePPI'].round(2)

# Criar o gráfico com Plotly Express
fig = px.bar(dados_grafico,
             x=dados_grafico.index,
             y=['PPI', 'CorePPI'],
             barmode='group',
             labels={'value': 'Variação Percentual (%)', 'variable': 'Índice'},
             text_auto=True,  # Exibir os valores arredondados
             color_discrete_map={'PPI': 'blue', 'CorePPI': 'darkorange'},
             title="PPI e CorePPI (Variação Mensal)")  # Adiciona o título

# Formatar o eixo x para mostrar ano e mês
fig.update_xaxes(
    tickformat="%Y-%m",  # Formato para ano e mês (YYYY-MM)
    dtick="M1",          # Intervalo de ticks mensal
    title_text="Data (Ano-Mês)"  # Título do eixo x
)

# Ajustar a posição e o tamanho da fonte dos números acima das barras
fig.update_traces(textposition='outside', textfont_size=25)

# Personalizar o formato do texto para incluir o símbolo de porcentagem
fig.update_traces(texttemplate='%{y:.2f}%')

# Exibir o gráfico
fig.show()
```

### IPC and Core IPC (YoY Comparison)

```python
import pandas_datareader.data as web
import datetime
import plotly.express as px

start = datetime.datetime(2000, 1, 1)
end = datetime.datetime.now()

cpi = web.DataReader('CPIAUCSL', 'fred', start, end)
core_cpi = web.DataReader('CPILFESL', 'fred', start, end)

# Calculate YoY percentage change
cpi['YoY_Change'] = cpi['CPIAUCSL'].pct_change(periods=12) * 100
core_cpi['YoY_Change'] = core_cpi['CPILFESL'].pct_change(periods=12) * 100

# Combine data for plotting
yoy_data = cpi[['YoY_Change']].rename(columns={'YoY_Change': 'CPI'})
yoy_data['Core CPI'] = core_cpi['YoY_Change']

# Create the chart with specified colors
fig = px.line(yoy_data,
              x=yoy_data.index,
              y=['CPI', 'Core CPI'],
              title='CPI and Core CPI YoY Comparison',
              color_discrete_map={'CPI': 'blue', 'Core CPI': 'darkorange'})

# Add a dotted line at 2%
fig.add_shape(type='line',
              x0=yoy_data.index.min(),
              y0=2,
              x1=yoy_data.index.max(),
              y1=2,
              line=dict(color='black', dash='dot'))

# Add a solid line at 0%
fig.add_shape(type='line',
              x0=yoy_data.index.min(),
              y0=0,
              x1=yoy_data.index.max(),
              y1=0,
              line=dict(color='black'))

fig.update_layout(yaxis_title='YoY Percentage Change (%)',
                  xaxis_title='Date')
fig.show()
```

### Core PCE (3M E 6M Annualized)

```python
import pandas_datareader.data as web
import datetime
import plotly.express as px

start = datetime.datetime(2015, 1, 1)
end = datetime.datetime.now()

core_pce = web.DataReader('PCEPILFE', 'fred', start, end)

# Calcula a variação percentual
core_pce['3-Month_Change'] = core_pce['PCEPILFE'].pct_change(periods=3)
core_pce['6-Month_Change'] = core_pce['PCEPILFE'].pct_change(periods=6)

# Anualiza a variação percentual
core_pce['3-Month_Annualized'] = (1 + core_pce['3-Month_Change']) ** (12 / 3) - 1
core_pce['6-Month_Annualized'] = (1 + core_pce['6-Month_Change']) ** (12 / 6) - 1


# Cria o gráfico com cores especificadas
fig = px.line(core_pce,
              x=core_pce.index,
              y=['3-Month_Annualized', '6-Month_Annualized'],
              title='Core PCE Anualizado em 3 e 6 Meses',
              color_discrete_map={'3-Month_Annualized': 'blue', '6-Month_Annualized': 'darkorange'})

# Adiciona uma linha sólida em 0%
fig.add_shape(type='line',
              x0=core_pce.index.min(),
              y0=0,
              x1=core_pce.index.max(),
              y1=0,
              line=dict(color='black'))

# Adiciona uma linha pontilhada preta em 0.02
fig.add_shape(type='line',
              x0=core_pce.index.min(),
              y0=0.02,
              x1=core_pce.index.max(),
              y1=0.02,
              line=dict(color='black', dash='dot'))  # dash='dot' cria a linha pontilhada


fig.update_layout(yaxis_title='Variação Percentual Anualizada (%)',
                  xaxis_title='Data',
                 legend_title_text='')  # Remove o título da legenda

fig.show()
```
